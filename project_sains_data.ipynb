```python
def kalkulator(op, num1, num2):
    if op == '+':
        return num1 + num2
    elif op == '-':
        return num1 - num2
    elif op == '*':
        return num1 * num2
    elif op == '/':
        if num2 == 0:
            print("!! num2 tidak boleh bernilai")
            return 0
        else:
            return num1 / num2
    else:
        print("!! Operator tidak Valid !!")
        return 0

def main():
    ketik = True
    print(f"============Kalkulator Python============\n")
    while ketik:
        print(">> pilih Operator: ")
        op = input(">> ")
        print(">> Masukan num1: ")
        num1 = int(input(">> "))
        print(">> Masukan num2: ")
        num2 = int(input(">> "))
        hasil = kalkulator(op, num1, num2)
        print(f"#JUMLAH# : {hasil}\n")
        pilih = input(">> Mau lanjut? (y/n): ")
        if pilih == 'n':
            ketik = False

main()
```
